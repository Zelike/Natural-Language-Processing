### 任务1：安装 Miniconda

1. 从官网下载 Miniconda3：

### 任务4：安装 NLP 常用库

### 任务6：环境验证

新建 Notebook，运行以下代码：


```python
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn
import jieba
import nltk

print('Python:', sys.version)
print('numpy:', np.__version__)
print('pandas:', pd.__version__)
print('sklearn:', sklearn.__version__)
print('jieba:', jieba.__version__)
print('nltk:', nltk.__version__)
```

### 任务7：第一个 NLP 小实验——中文分词与词频统计


```python
import jieba
from collections import Counter

text = '自然语言处理是人工智能皇冠上的明珠，Python是自然语言处理常用的编程工具。'
words = jieba.lcut(text)
print('分词结果：', words)

word_freq = Counter(words)
print('词频统计：', word_freq.most_common(10))
```

### 任务8：英文分词验证


```python
import nltk
nltk.download('punkt')
from nltk.tokenize import word_tokenize

sentence = 'Natural language processing is a fascinating field.'
print(word_tokenize(sentence))
```

若 `punkt` 下载失败，可尝试：

    nltk.download('punkt_tab')

或由教师提供离线 `nltk_data` 包。

## 五、一键检查脚本

新建 `check_env.py`：


```python
import sys

print('Python 版本：', sys.version)

mods = ['numpy', 'pandas', 'matplotlib', 'sklearn', 'jieba', 'nltk']
for m in mods:
    try:
        mod = __import__(m)
        print(f'{m}: {getattr(mod, "__version__", "OK")}')
    except Exception as e:
        print(f'{m}: 失败 -> {e}')

import jieba
text = '自然语言处理是人工智能皇冠上的明珠'
print('jieba分词：', jieba.lcut(text))
```

运行：

    python check_env.py

能正确输出版本和分词结果，即视为环境搭建成功。


## 六、常见问题与解决

| 问题 | 解决办法 |
|---|---|
| `conda` 不是内部或外部命令 | 使用 Anaconda Prompt，或执行 `conda init` 后重启终端 |
| pip 下载慢 | 配置清华源：`pip config set global.index-url https://pypi.tuna.tsinghua.edu.cn/simple` |
| Jupyter 内核不对 | 执行 `python -m ipykernel install --user --name nlp --display-name "Python (NLP)"` |
| `jieba` 导入失败 | 确认已 `conda activate nlp`，再 `pip install jieba` |
| 端口被占用 | `jupyter notebook --port 8889` |
| 中文绘图乱码 | `plt.rcParams['font.sans-serif'] = ['SimHei']` |
| 磁盘空间不足 | `conda clean -a` 清理缓存 |
